## **Business context**
An e-commerce data set from a real-world organisation to perform customer segmentation with clustering models to improve marketing efforts (SAS, 2024). It is a transnational data set with customers from five continents (Oceania, North America, Europe, Africa, and Asia) and 47 countries.

The data set contains 951,668 rows, each representing a product a customer ordered. As each customer is unique, it is critical to identify and/or create new features for customer segmentation to inform marketing efforts. The data set has 20 features
- **Quantity:** The quantity the customer orders (e.g. 1, 2, 3).
- **City:** Name of the customer's residence (e.g. Leinster, Berowra, Northbridge).
- **Continent:** Name of the continent where the customer resides (Oceania, North America).
- **Postal code:** Where the customer resides (e.g. 6437, 2081, 2063).   
- **State province:** State or province where the customer resides (e.g. Western Australia, Quebec, New South Wales).
- **Order date:** The date the order was placed (e.g. 1 January 2012, 20 June 2014).
- **Delivery date:** The date the order was delivered (e.g. 12 April 2014, 19 November 2016).

- **Discount:** Percentage or normal total retail price (e.g. 50%, 30%).
- **Order type label:** Method in which the order was placed (e.g. internet sale, retail sale).

- **Customer type:** Loyalty member level (e.g. internet/catalogue customers, Orion club gold members high activity).
- **Order ID:** Unique order identifier (e.g. 1230000033).
- **Profit:** Total profit is calculated: $Total\:profit=Total\:revenue-(Unit\:cost\times Quantity)$.
- **Days to delivery:** The number of days for delivery is calculated: $Delivery\:days=Delivery\:date-Order\:date$ (e.g. 6, 3, 2).
- **Loyalty number:** Loyal customer (99) versus non-loyal customer (0).
- **Customer ID:** A unique identifier for the customer (e.g. 8818, 47793).

Since we have a transnational data set, which implies customers from different continents, several metrics are important when performing customer segmentation for target marketing. From a marketing perspective, the following five metrics help to understand the nuance of the customer base, buying behaviour, preferences, and value to the business.
- **Frequency** indicates how often a customer purchases over a given period of time. A high frequency indicates a loyal customer, a high level of satisfaction, trust or brand loyalty, and/or effective marketing efforts. Frequency based on purchases guides a business in the effectiveness of target marketing campaigns and how to target less active customers.
- **Recency** measures how recently a customer made a purchase or placed an order. It helps predict customer churn (turnover) and engagement. A customer is a business’s most valuable asset, so securing customer retention is essential. As a result, recent customer activity can indicate their overall satisfaction and engagement.
- **Customer lifetime value (CLV)** indicates the average or total value a customer contributes to a business over the course of their relationship. In other words, CLV is a metric of the total income a business can expect to generate from a customer as long as said customer remains a loyal client. CLV helps to prioritise marketing efforts and resources as it focuses on customers who are expected to bring the most value over time. Therefore, retaining high-value customers.
- The **average unit cost** indicates if the customer prefers low cost or high cost items. This is related to the profitability of purchases. Customers buying products with a higher average unit cost price should be targeted differently. Customer segmentation assists in identifying these customers.
- **Customer age** indicates the number of years a customer has lived.

```python
# Profit column shows negative values in accounting format, ($0.40).
# This column must be handled differently to "Total Revenue", "Unit Cost"

# Profit: handle $ and negatives i.e. ($0.40)
s = data_clean["Profit"].astype(str).str.strip() #removes unwanted whitespace

# Identify values wrapped in brackets
neg = s.str.startswith("(") & s.str.endswith(")")

# Strip $, commas, brackets and spaces, then convert to numeric
s = s.str.replace(r"[$,()\s]", "", regex=True)
data_clean["Profit"] = pd.to_numeric(s, errors="coerce")

# Apply the negative sign to the bracketed values
data_clean.loc[neg, "Profit"] *= -1

# Verify
print(data_clean["Profit"].isna().sum(), "NaNs")
print((data_clean["Profit"] < 0).sum(), "rows with negative profit")
print(data_clean["Profit"].describe())
```

##Feature engineering:
 - Create new features for frequency, recency, CLV, average unit cost, and customer age.
 - Perform feature scaling and encoding if needed.

 These five features wil be the input for the clustering model.

 **Average unit cost**: indication of customer preference for low or high cost times. Related to profitability of purchases. Customers buying products a with higher average unit cost are a group of interest

 **Frequency**: indicates how often a customer purchases over a given period. A high frequency indicates a loyal customer, high level of satisfaction, trust or brand loyalty, and/or efficient marketing efforts.

 **Customer age**: no. of years a customer has lived

 **Recency**: measures how recently a customr made a purchase. It helps predict customer churn and engagement.

 **Customer lifetime value (CLV)**: indicates the average or total value a customer contributes to a business over the course of their relationship. A metric of total income a business can expect to generate from a customer.

```python
# Average unit cost included in aggregated data - 'Avg_Unit_Cost'

# Frequency (no.of distinct orders per customer)
customers["Frequency"] = customers["Num_Orders"]

# Customer age
customers["Customer_Age"] = (snapshot_date - customers["Birth_Date"]).dt.days // 365.25

# Recency
customers["Recency"] = (snapshot_date - customers["Last_Order"]).dt.days

# CLV
customers["CLV"] = customers["Total_Revenue"] / customers["Num_Orders"]

```

# Clustering with ML models

## K-means clustering

```python
# Perform k-means clustering using scikit-learn
from sklearn.cluster import KMeans

# Defining what K-means is allowed to see
X_df_features = ["Frequency", "Recency", "CLV", "Avg_Unit_Cost", "Customer_Age"]

# Creates the model
# n_init=10 means ten complete runs from different starts, keeping the lowest inertia
# random_state fixes the randomness
kmeans = KMeans(n_clusters=6, random_state=42, n_init=10)

# fit() runs the steps on the five transformed features
# predict() assigns rows to the nearest learned centroid
# fit_predict runs the assign-and-update loop on 68,300 rows, then returns a list of 68,300 cluster numbers (0, 1 or 2). Assigning them to a new column, Cluster gives every customer a label.
X_df["Cluster"] = kmeans.fit_predict(X_df[X_df_features])

# Extracting centroids
# Pulls out the final centre of each cluster, which is the mean of each feature for that cluster. Its shape (3, 5) means 3 clusters × 5 features.
centroids = kmeans.cluster_centers_


#After fitting cluster_centers holds the 3x5 centroids
print("Final Centroids:", centroids)

# Values are standard deviations from the overall average
print(kmeans.cluster_centers_.shape) # should be (6, 5)
```

```python
inertia_ = []
for k in range(2, 11):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10) #init='k-means++' is default in scikit learn so not added
    kmeans.fit(X_df[X_df_features])
    inertia_.append(kmeans.inertia_)
```

```python
# Elbow method to determine the number of clusters


plt.plot(range(2, 11), inertia_, marker='o')
plt.title('Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('Inertia')
plt.show()
```

The significance of the elbow method lies in identifying the point on the plot where the rate of decrease in within-cluster variation starts to diminish, forming a distinct bend or elbow. The bend represents the optimal number of clusters, striking a balance between capturing meaningful patterns in the data and avoiding unnecessary complexity.

The elbow here is soft and the bend point is at k=6.

```python
# Silhouette score
from sklearn.metrics import silhouette_samples, silhouette_score

ks = range(2, 11)
sil_ = []

for k in ks:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_df[X_df_features])
    sil_.append(silhouette_score(X_df[X_df_features], labels, sample_size=10000, random_state=42))

plt.plot(ks, sil_, marker="o")
plt.title("Silhouette score by k")
plt.xlabel("Number of clusters (k)")
plt.ylabel("Average silhouette score")
plt.show()
```

```python
print("For 2-10 clusters, the average silhouette_score is :", sil_)
```

The Silhouette scores are between 0.212 and 0.258. Silhouette values close to 0 means customers are spread continuously and not in separated groups. This agrees with the feature pairplots shown in the EDA.

The elbow method showed the largest reduction in inertia at k=6, with diminishing returns afterwards. Silhouette scores are chosn at the seond peak for 0.243.

Silhouette sample used to break down the average further providing a silhouette coefficient for each data point in the data set. This helps understand the distribution of the sihouette coefficients across all data points.

```python
k = 2
sample = X_df.sample(10000, random_state=42)
Xs = sample[X_df_features]

km = KMeans(n_clusters=k, random_state=42, n_init=10)
labels = km.fit_predict(Xs)

# one silhouette coefficient per customer
sil_vals = silhouette_samples(Xs, labels)
# single average for all (red dashed line)
avg = silhouette_score(Xs, labels)

fig, ax = plt.subplots(figsize=(8, 6))
y_lower = 10
for i in range(k):
    vals = np.sort(sil_vals[labels == i]) #selects only the sil_vals of customer and sorts, labels==i gives True/False for customer and picks matching values
    # works out where the block ends. One row of height per customer, so a cluster with 10,000 customers takes 10,000 units of height. That’s why bigger clusters appear thicker.
    y_upper = y_lower + len(vals)
    ax.fill_betweenx(np.arange(y_lower, y_upper), 0, vals, alpha=0.7) #colours the area
    ax.text(-0.05, y_lower + 0.5 * len(vals), str(i))
    y_lower = y_upper + 10 # moves the starting point up for the next cluster, leaving a gap of 10 units so the blocks don’t touch.

ax.axvline(avg, color="red", linestyle="--")
ax.set_xlabel("Silhouette coefficient")
ax.set_ylabel("Cluster")
ax.set_yticks([])
ax.set_title(f"Silhouette plot for k = {k} (average = {avg:.2f})")
plt.show()
```

##Hierarchical clustering with a dendrogram

####Linkage method consideration (single, complete, average, centroid, ward)


```python
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

sample = X_df[X_df_features].sample(3000, random_state=42)

methods = ["ward", "complete", "average", "single", "centroid"]

fig, axes = plt.subplots(2, 3, figsize=(16, 8))
axes = axes.ravel()

for ax, method in zip(axes, methods):
    Z_m = linkage(sample, method=method)
    dendrogram(Z_m, truncate_mode="lastp", p=20, no_labels=True, ax=ax)
    ax.set_title(method)

axes[-1].axis("off")   # hide the unused sixth panel

plt.tight_layout()
plt.show()
```

```python
from scipy.cluster.hierarchy import fcluster
# Polulation of cluster sizes for each linkage method
# k=5
rows = []
for method in methods:
    Z_m = linkage(sample, method=method)
    sizes = pd.Series(fcluster(Z_m, t=5, criterion="maxclust")).value_counts().sort_index().tolist()
    rows.append({"method": method, "cluster_sizes": sizes})

print(pd.DataFrame(rows))
```

# Reflect

The K-means (k=6) clustering model segmented 68,300 customers into distinct interpretable behavioural profiles that can be considered for targeted marketing strategies. The choice of k was supported by the elbow at 6 clusters, silhouette score of 0.24, and a ward dendrogram consistent with six branches. The interpretability of the clusters was supported by the cluster distribution in the feature box plots, while PCA and t-SNE showed segments representing the clusters.
The clusters, although defined, were not isolated separate groups suggesting that customer behaviour varies continuously. The soft elbow and the low silhouette scores implied overlapping clusters. K-means is sensitive to skewed features, and extreme values may contribute to this because distance-based measures are affected.  This can pose a challenge for customer clustering as extreme values of features may represent valuable customer groups that you may not want to remove. This is demonstrated by the small cluster of about 785 customers (1%) having the highest CLV and unit cost, and it exists because those extreme values were kept.
Furthermore, CLV and Average_Unit_Cost are strongly correlated (r =0.73), so spending level may have carried more weight in the clustering profiling than other features. Further exploration could investigate the robustness of the clusters by removing one of the correlated features and investigating other features that may affect customer behaviour.
